# HSP molecule editor

Run the first code cell to open ethanol in the editor. Drag atoms or use the editor controls, then run the second code cell to retrieve the edited coordinates.

HSP is already installed in this Binder environment.

```python
import hylleraas as hsp
from IPython.display import display

molecule = hsp.Molecule("CCO")  # Ethanol
editor = molecule.view(viewer="editor")
display(editor)
```

```python
edited_molecule = editor.to_molecule()
print(edited_molecule.xyz_string)
```

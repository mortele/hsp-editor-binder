### HSP molecule editor beta
All necessary packages are installed in this environment. 
Run the imports, then choose either the empty-editor cell to build from scratch or the ethanol cell (built from the SMILES string `"CCO"`).

The last cell prints the `.xyz` contents of the edited molecule. `editor` refers to whichever editor you opened most recently; run the export cell again after editing to fetch its current geometry. Close unused editors with their **X** button.

```python
import hylleraas as hsp
from IPython.display import display
```

```python
editor = hsp.edit_molecule()
display(editor)
```

```python
ethanol = hsp.Molecule("CCO")
editor = hsp.edit_molecule(ethanol)
display(editor)
```

```python
edited_molecule = editor.to_molecule()
print(edited_molecule.xyz_string)
```
